## Supply-chain diagnostics

The following analyses are exploratory diagnostics used to understand which components and suppliers contribute to the vehicle-level field defects. They are not separate final datasets.


```python
def combine_defect_flags(*series):
    values = pd.concat(series, axis=1)

    defective = values.eq(1).any(axis=1)
    unknown = values.isna().any(axis=1)

    result = pd.Series(
        0,
        index=values.index,
        dtype="Int8"
    )

    result = result.mask(
        ~defective & unknown,
        pd.NA
    )

    result = result.mask(
        defective,
        1
    )

    return result
```

```python
family = "K3SG1"

component = pd.read_csv(
    KOMPONENTE_FOLDER / "Komponente_K3SG1.csv",
    usecols=[
        "ID",
        "Werksnummer",
        "Fehlerhaft"
    ],
    dtype="string"
)

bom = pd.read_csv(
    KOMPONENTE_FOLDER
    / "Bestandteile_Komponente_K3SG1.csv",
    sep=";",
    usecols=[
        "ID_K3SG1",
        "ID_T21",
        "ID_T22",
        "ID_T23"
    ],
    dtype="string"
)

if not component["ID"].is_unique:
    raise ValueError(
        "Duplicate K3SG1 IDs in component file"
    )

if not bom["ID_K3SG1"].is_unique:
    raise ValueError(
        "Duplicate K3SG1 IDs in BOM"
    )

component_lookup = component.set_index("ID")

k3sg1_quality = pd.DataFrame({
    "ID": bom["ID_K3SG1"],
    "Tier1_Werksnummer":
        bom["ID_K3SG1"].map(
            component_lookup["Werksnummer"]
        ),
    "Fehlerhaft_Komponente_Roh":
        pd.to_numeric(
            bom["ID_K3SG1"].map(
                component_lookup["Fehlerhaft"]
            ),
            errors="coerce"
        ).astype("Int8")
})

for part in ["T21", "T22", "T23"]:
    part_data = pd.read_csv(
        EINZELTEIL_FOLDER
        / f"Einzelteil_{part}.csv",
        usecols=[
            "ID",
            "Werksnummer",
            "Fehlerhaft"
        ],
        dtype="string"
    )

    if not part_data["ID"].is_unique:
        raise ValueError(
            f"Duplicate {part} IDs"
        )

    part_lookup = part_data.set_index("ID")

    part_id_column = f"ID_{part}"

    k3sg1_quality[
        f"Tier2_Werksnummer_{part}"
    ] = bom[part_id_column].map(
        part_lookup["Werksnummer"]
    )

    k3sg1_quality[
        f"Fehlerhaft_{part}"
    ] = pd.to_numeric(
        bom[part_id_column].map(
            part_lookup["Fehlerhaft"]
        ),
        errors="coerce"
    ).astype("Int8")

    del part_data
    del part_lookup

k3sg1_quality[
    "Fehlerhaft_Einzelteil"
] = combine_defect_flags(
    k3sg1_quality["Fehlerhaft_T21"],
    k3sg1_quality["Fehlerhaft_T22"],
    k3sg1_quality["Fehlerhaft_T23"]
)

k3sg1_quality[
    "Anzahl_Fehlerhafte_Einzelteile"
] = (
    k3sg1_quality[
        [
            "Fehlerhaft_T21",
            "Fehlerhaft_T22",
            "Fehlerhaft_T23"
        ]
    ]
    .eq(1)
    .sum(axis=1)
    .astype("Int8")
)

k3sg1_quality[
    "Fehlerhaft_Komponente"
] = combine_defect_flags(
    k3sg1_quality[
        "Fehlerhaft_Komponente_Roh"
    ],
    k3sg1_quality[
        "Fehlerhaft_Einzelteil"
    ]
)

k3sg1_quality.head()
```

```python
def family_role(family):
    if family in {"K4", "K6"}:
        return "Karosserie"
    if family.startswith("K3"):
        return "Schaltung"
    if family.startswith("K2"):
        return "Sitze"
    if family.startswith("K1"):
        return "Motor"
    raise ValueError(f"Unknown family: {family}")


def process_family(family, vehicle):
    role = family_role(family)

    parts = component_parts[family]

    bom_part_columns = [
        f"ID_T{int(part[1:])}"
        for part in parts
    ]

    component_id_column = f"ID_{family}"

    component = pd.read_csv(
        KOMPONENTE_FOLDER
        / f"Komponente_{family}.csv",
        usecols=[
            "ID",
            "Werksnummer",
            "Fehlerhaft"
        ],
        dtype="string"
    )

    bom = pd.read_csv(
        KOMPONENTE_FOLDER
        / f"Bestandteile_Komponente_{family}.csv",
        sep=";",
        usecols=[
            component_id_column,
            *bom_part_columns
        ],
        dtype="string"
    )

    if not component["ID"].is_unique:
        raise ValueError(
            f"{family}: duplicate component IDs"
        )

    if not bom[component_id_column].is_unique:
        raise ValueError(
            f"{family}: duplicate BOM component IDs"
        )

    component_lookup = component.set_index("ID")

    quality = pd.DataFrame({
        "ID": bom[component_id_column],
        "Tier1_Werksnummer":
            bom[component_id_column].map(
                component_lookup["Werksnummer"]
            ),
        "Fehlerhaft_Roh":
            pd.to_numeric(
                bom[component_id_column].map(
                    component_lookup["Fehlerhaft"]
                ),
                errors="coerce"
            ).astype("Int8")
    })

    part_defect_columns = []

    for part, bom_part_column in zip(
        parts,
        bom_part_columns
    ):
        part_data = pd.read_csv(
            EINZELTEIL_FOLDER
            / f"Einzelteil_{part}.csv",
            usecols=[
                "ID",
                "Fehlerhaft"
            ],
            dtype="string"
        )

        if not part_data["ID"].is_unique:
            raise ValueError(
                f"{part}: duplicate IDs"
            )

        part_lookup = (
            part_data
            .set_index("ID")["Fehlerhaft"]
        )

        defect_column = f"Fehlerhaft_{part}"

        quality[defect_column] = pd.to_numeric(
            bom[bom_part_column].map(
                part_lookup
            ),
            errors="coerce"
        ).astype("Int8")

        part_defect_columns.append(
            defect_column
        )

        del part_data
        del part_lookup

    quality[
        "Fehlerhaft_Einzelteil"
    ] = combine_defect_flags(
        *[
            quality[column]
            for column in part_defect_columns
        ]
    )

    quality[
        "Anzahl_Fehlerhafte_Einzelteile"
    ] = (
        quality[
            part_defect_columns
        ]
        .eq(1)
        .sum(axis=1)
        .astype("Int8")
    )

    quality[
        "Fehlerhaft_Komponente"
    ] = combine_defect_flags(
        quality["Fehlerhaft_Roh"],
        quality["Fehlerhaft_Einzelteil"]
    )

    lookup = quality.set_index("ID")

    vehicle_id_column = f"ID_{role}"

    mask = (
        vehicle[vehicle_id_column]
        .astype("string")
        .str.startswith(f"{family}-")
    )

    vehicle.loc[
        mask,
        f"Tier1_Werksnummer_{role}"
    ] = (
        vehicle.loc[
            mask,
            vehicle_id_column
        ]
        .map(
            lookup["Tier1_Werksnummer"]
        )
    )

    vehicle.loc[
        mask,
        f"Fehlerhaft_{role}_Roh"
    ] = (
        vehicle.loc[
            mask,
            vehicle_id_column
        ]
        .map(
            lookup["Fehlerhaft_Roh"]
        )
    )

    vehicle.loc[
        mask,
        f"Fehlerhaft_{role}_Einzelteil"
    ] = (
        vehicle.loc[
            mask,
            vehicle_id_column
        ]
        .map(
            lookup["Fehlerhaft_Einzelteil"]
        )
    )

    vehicle.loc[
        mask,
        f"Anzahl_Fehlerhafte_Einzelteile_{role}"
    ] = (
        vehicle.loc[
            mask,
            vehicle_id_column
        ]
        .map(
            lookup[
                "Anzahl_Fehlerhafte_Einzelteile"
            ]
        )
    )

    vehicle.loc[
        mask,
        f"Fehlerhaft_{role}"
    ] = (
        vehicle.loc[
            mask,
            vehicle_id_column
        ]
        .map(
            lookup["Fehlerhaft_Komponente"]
        )
    )

    result = {
        "family": family,
        "role": role,
        "produced_components": len(quality),
        "installed_components": int(mask.sum()),
        "missing_tier1": int(
            vehicle.loc[
                mask,
                f"Tier1_Werksnummer_{role}"
            ].isna().sum()
        ),
        "missing_defect": int(
            vehicle.loc[
                mask,
                f"Fehlerhaft_{role}"
            ].isna().sum()
        )
    }

    del component
    del bom
    del component_lookup
    del quality
    del lookup

    return result
```

```python
family_results = []

for family in sorted(type11_families):
    print(f"Processing Type11 {family}...")

    result = process_family(
        family,
        vehicle_11
    )

    family_results.append(result)

    print(
        f"{family}: "
        f"{result['installed_components']:,} installed, "
        f"{result['missing_tier1']} missing Tier1, "
        f"{result['missing_defect']} missing defects"
    )

for family in sorted(type21_families):
    print(f"Processing Type21 {family}...")

    result = process_family(
        family,
        vehicle_21
    )

    family_results.append(result)

    print(
        f"{family}: "
        f"{result['installed_components']:,} installed, "
        f"{result['missing_tier1']} missing Tier1, "
        f"{result['missing_defect']} missing defects"
    )

family_results = pd.DataFrame(
    family_results
)

display(family_results)
```

```python
tier1_lookup = (
    tier1_werke
    .set_index(
        "Tier1_Werksnummer"
    )
)

for role in roles:
    plant_column = (
        f"Tier1_Werksnummer_{role}"
    )

    vehicle_quality[
        f"ORT_Tier1_{role}"
    ] = (
        vehicle_quality[
            plant_column
        ]
        .map(
            tier1_lookup[
                "ORT_Tier1"
            ]
        )
    )

    vehicle_quality[
        f"PLZ_Tier1_{role}"
    ] = (
        vehicle_quality[
            plant_column
        ]
        .map(
            tier1_lookup[
                "PLZ_Tier1"
            ]
        )
    )

    vehicle_quality[
        f"Breitengrad_Tier1_{role}"
    ] = pd.to_numeric(
        vehicle_quality[
            plant_column
        ]
        .map(
            tier1_lookup[
                "Breitengrad_Tier1"
            ]
        ),
        errors="coerce"
    )

    vehicle_quality[
        f"Laengengrad_Tier1_{role}"
    ] = pd.to_numeric(
        vehicle_quality[
            plant_column
        ]
        .map(
            tier1_lookup[
                "Laengengrad_Tier1"
            ]
        ),
        errors="coerce"
    )

print(
    "Missing Tier-1 locations:",
    sum(
        vehicle_quality[
            f"ORT_Tier1_{role}"
        ].isna().sum()
        for role in roles
    )
)

del tier1_lookup
```

```python
tier1_quality_rows = []

for role in roles:
    plant_column = f"Tier1_Werksnummer_{role}"

    summary = (
        vehicle_quality
        .groupby(
            [
                "OEM",
                "Werksnummer_OEM",
                "ORT_OEM",
                plant_column
            ],
            as_index=False
        )
        .agg(
            Einheiten=(
                f"ID_{role}",
                "size"
            ),
            Fehlerhaft=(
                f"Fehlerhaft_{role}",
                "sum"
            ),
            Direkte_Fehler=(
                f"Fehlerhaft_{role}_Roh",
                "sum"
            )
        )
        .rename(
            columns={
                plant_column:
                    "Tier1_Werksnummer"
            }
        )
    )

    summary["Typ"] = role

    summary["Fehlerrate"] = (
        summary["Fehlerhaft"]
        / summary["Einheiten"]
    )

    summary["Direkte_Fehlerrate"] = (
        summary["Direkte_Fehler"]
        / summary["Einheiten"]
    )

    tier1_quality_rows.append(summary)

tier1_quality = pd.concat(
    tier1_quality_rows,
    ignore_index=True
)

del tier1_quality_rows

tier1_quality = tier1_quality.merge(
    tier1_werke,
    how="left",
    on="Tier1_Werksnummer",
    validate="many_to_one"
)

tier1_quality = (
    tier1_quality
    .sort_values(
        "Fehlerrate",
        ascending=False
    )
    .reset_index(drop=True)
)

display(tier1_quality)
```
