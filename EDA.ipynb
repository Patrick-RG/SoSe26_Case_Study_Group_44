```python
from pathlib import Path
import pandas as pd
import csv

COMPONENT_FOLDER = Path("data/Komponente")


def detect_separator(path):
    with open(path, "r", encoding="utf-8", errors="replace") as f:
        first_line = f.readline()

    candidates = [",", ";", "\t", "|"]

    return max(
        candidates,
        key=lambda sep: first_line.count(sep)
    )


for family in component_families:

    component_path = (
        COMPONENT_FOLDER
        / f"Komponente_{family}.csv"
    )

    bom_path = (
        COMPONENT_FOLDER
        / f"Bestandteile_Komponente_{family}.csv"
    )

    print("\n" + "=" * 70)
    print(family)

    for label, path in [
        ("COMPONENT", component_path),
        ("COMPONENT BOM", bom_path)
    ]:

        sep = detect_separator(path)

        df_test = pd.read_csv(
            path,
            sep=sep,
            nrows=3
        )

        print(f"\n{label}")
        print("separator:", repr(sep))
        print("columns:")
        print(df_test.columns.tolist())

        display(df_test)
```

```python
import pandas as pd


def component_id_column(family):
    if family.startswith("K1"):
        return "ID_Motor"
    elif family.startswith("K2"):
        return "ID_Sitze"
    elif family.startswith("K3"):
        return "ID_Schaltung"
    elif family.startswith("K4") or family.startswith("K6"):
        return "ID_Karosserie"

    raise ValueError(f"Unknown component family: {family}")


def normalize_component_chunk(df, family):

    id_col = component_id_column(family)

    fields = [
        id_col,
        "Produktionsdatum",
        "Herstellernummer",
        "Werksnummer",
        "Fehlerhaft",
        "Fehlerhaft_Datum",
        "Fehlerhaft_Fahrleistung",
    ]

    clean = pd.DataFrame(index=df.index)

    # Case 1: .x/.y or .x/.y/base
    if f"{id_col}.x" in df.columns:

        possible_id_cols = [
            f"{id_col}.x",
            f"{id_col}.y",
            id_col
        ]

        possible_id_cols = [
            col for col in possible_id_cols
            if col in df.columns
        ]

        populated = df[possible_id_cols].notna().sum(axis=1)

        if (populated > 1).any():
            raise ValueError(
                f"{family}: some rows contain multiple "
                "component records (.x/.y/base)"
            )

        if (populated == 0).any():
            raise ValueError(
                f"{family}: some rows contain no component ID"
            )

        for field in fields:

            candidates = [
                f"{field}.x",
                f"{field}.y",
                field
            ]

            candidates = [
                col for col in candidates
                if col in df.columns
            ]

            result = df[candidates[0]]

            for col in candidates[1:]:
                result = result.combine_first(df[col])

            clean[field] = result

    # Case 2: already normal
    else:

        for field in fields:
            if field in df.columns:
                clean[field] = df[field]

        # Special production-date format:
        # days since origin date
        if (
            "Produktionsdatum" not in df.columns
            and
            "Produktionsdatum_Origin_01011970" in df.columns
            and
            "origin" in df.columns
        ):
            days = pd.to_numeric(
                df["Produktionsdatum_Origin_01011970"],
                errors="coerce"
            )

            origin = pd.to_datetime(
                df["origin"],
                format="%d-%m-%Y",
                errors="coerce"
            )

            clean["Produktionsdatum"] = (
                origin
                + pd.to_timedelta(days, unit="D")
            )

    # Standard names

    clean = clean.rename(
        columns={id_col: "ID"}
    )

    clean.insert(
        0,
        "Komponententyp",
        family
    )

    # Standard data types

    clean["ID"] = clean["ID"].astype("string")

    clean["Herstellernummer"] = (
        pd.to_numeric(
            clean["Herstellernummer"],
            errors="coerce"
        )
        .astype("Int64")
        .astype("string")
    )

    clean["Werksnummer"] = (
        pd.to_numeric(
            clean["Werksnummer"],
            errors="coerce"
        )
        .astype("Int64")
        .astype("string")
    )

    clean["Produktionsdatum"] = pd.to_datetime(
        clean["Produktionsdatum"],
        errors="coerce"
    )

    clean["Fehlerhaft_Datum"] = pd.to_datetime(
        clean["Fehlerhaft_Datum"],
        errors="coerce"
    )

    clean["Fehlerhaft"] = pd.to_numeric(
        clean["Fehlerhaft"],
        errors="coerce"
    ).astype("Int8")

    clean["Fehlerhaft_Fahrleistung"] = pd.to_numeric(
        clean["Fehlerhaft_Fahrleistung"],
        errors="coerce"
    ).astype("Float64")

    return clean[
        [
            "Komponententyp",
            "ID",
            "Produktionsdatum",
            "Herstellernummer",
            "Werksnummer",
            "Fehlerhaft",
            "Fehlerhaft_Datum",
            "Fehlerhaft_Fahrleistung",
        ]
    ]
```

```python
for family in component_families:

    path = (
        COMPONENT_FOLDER
        / f"Komponente_{family}.csv"
    )

    sep = detect_separator(path)

    raw = pd.read_csv(
        path,
        sep=sep,
        nrows=5
    )

    clean = normalize_component_chunk(
        raw,
        family
    )

    print(
        family,
        "OK",
        "|",
        len(raw.columns),
        "->",
        len(clean.columns),
        "columns"
    )

    display(clean.head(2))
```

```python
component_summaries = {}

for family in component_families:

    input_path = (
        COMPONENT_FOLDER
        / f"Komponente_{family}.csv"
    )

    output_path = (
        COMPONENT_FOLDER
        / f"Komponente_{family}_clean.csv"
    )

    if output_path.exists():
        output_path.unlink()

    sep = detect_separator(input_path)

    total_rows = 0
    missing_id = 0
    duplicate_id = 0
    faulty_without_date = 0
    nonfaulty_with_date = 0

    first_chunk = True
    seen_ids = set()

    for raw in pd.read_csv(
        input_path,
        sep=sep,
        chunksize=50_000
    ):

        clean = normalize_component_chunk(
            raw,
            family
        )

        # Missing IDs
        missing_id += int(
            clean["ID"].isna().sum()
        )

        # Duplicate IDs within this chunk
        ids = clean["ID"].dropna()

        duplicate_id += int(
            ids.duplicated().sum()
        )

        # Duplicate IDs against previous chunks
        unique_ids = ids.unique()

        duplicate_id += sum(
            id_ in seen_ids
            for id_ in unique_ids
        )

        seen_ids.update(unique_ids)

        # Defect consistency
        faulty_without_date += int(
            (
                (clean["Fehlerhaft"] == 1)
                & clean["Fehlerhaft_Datum"].isna()
            ).sum()
        )

        nonfaulty_with_date += int(
            (
                (clean["Fehlerhaft"] == 0)
                & clean["Fehlerhaft_Datum"].notna()
            ).sum()
        )

        clean.to_csv(
            output_path,
            mode="a",
            header=first_chunk,
            index=False
        )

        first_chunk = False
        total_rows += len(clean)

        print(
            f"\r{family}: {total_rows:,} rows",
            end=""
        )

    component_summaries[family] = {
        "rows": total_rows,
        "missing_ID": missing_id,
        "duplicate_ID": duplicate_id,
        "faulty_without_date": faulty_without_date,
        "nonfaulty_with_date": nonfaulty_with_date,
    }

    print(
        f"\r{family}: finished — {total_rows:,} rows"
        + " " * 10
    )


component_summary_df = pd.DataFrame.from_dict(
    component_summaries,
    orient="index"
)

component_summary_df.index.name = "component"

display(component_summary_df)
```
